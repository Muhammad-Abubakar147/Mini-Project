**Importing Libraries**

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

**Load Dataset**

```python
zoo = pd.read_csv("zoo.csv")
classes = pd.read_csv("class.csv")
```

**Read Dataset**

```python
zoo.head()
```

```python
classes.head()
```

**Dataset information**

```python
zoo.info()
zoo.describe()
```

**Missing Values**

```python
zoo.isnull().sum()
```

**Merge Both dataset**

```python
# Merge zoo.csv with class.csv
zoo_data = pd.merge(
    zoo,
    classes,
    left_on="class_type",
    right_on="Class_Number"
)

# Check first 5 rows
zoo_data.head()
```

**Dataset Information**

```python
zoo_data.info()
```

**Animals Class distribution**

```python
plt.figure(figsize=(8,5))
sns.countplot(data=zoo_data, x="Class_Type")
plt.title("Number of Animals in Each Class")
plt.xticks(rotation=45)
plt.show()
```

**Domestic vs Wild Animals**

```python
plt.figure(figsize=(5,4))
sns.countplot(data=zoo_data, x="domestic")
plt.title("Domestic vs Wild Animals")
plt.show()
```

**Number of Legs**

```python
plt.figure(figsize=(7,4))
sns.countplot(data=zoo_data, x="legs")
plt.title("Distribution of Number of Legs")
plt.show()
```

**Heatmap (Correlation)**

```python
plt.figure(figsize=(12,8))
sns.heatmap(zoo_data.corr(numeric_only=True), cmap="coolwarm")
plt.title("Correlation Heatmap")
plt.show()
```

**Animals by Tail**

```python
plt.figure(figsize=(5,4))
sns.countplot(data=zoo_data, x="tail")
plt.title("Animals With and Without Tail")
plt.show()
```

**Animal Size**

```python
plt.figure(figsize=(5,4))
sns.countplot(data=zoo_data, x="catsize")
plt.title("Cat Size Distribution")
plt.show()
```
